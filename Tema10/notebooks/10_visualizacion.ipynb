## Primer gráfico

Script `primer_grafico.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

peticiones = pd.read_csv("Tema10/datos/peticiones.csv")
peticiones["fecha"] = pd.to_datetime(peticiones["fecha"])

# la figura es la imagen completa y los ejes son el área donde se dibuja
figura, ejes = plt.subplots(figsize=(8, 4))
ejes.plot(peticiones["fecha"], peticiones["peticiones_web"])
ejes.set_title("Peticiones web por día")
ejes.set_xlabel("Fecha")
ejes.set_ylabel("Peticiones")

figura.savefig("Tema10/graficos/peticiones_web.png")
print("Gráfico guardado en Tema10/graficos/peticiones_web.png")
```

## Líneas con dos series

Script `lineas_series.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

peticiones = pd.read_csv("Tema10/datos/peticiones.csv")
peticiones["fecha"] = pd.to_datetime(peticiones["fecha"])

figura, ejes = plt.subplots(figsize=(8, 4))

# cada llamada a plot() agrega una serie; label alimenta la leyenda
ejes.plot(peticiones["fecha"], peticiones["peticiones_web"], label="Web")
ejes.plot(peticiones["fecha"], peticiones["peticiones_api"], label="API")

ejes.set_title("Peticiones por día")
ejes.set_xlabel("Fecha")
ejes.set_ylabel("Peticiones")
ejes.set_ylim(0, 25000)  # el eje vertical empieza en cero
ejes.grid(True)
ejes.legend()

figura.savefig("Tema10/graficos/peticiones_series.png", dpi=150, bbox_inches="tight")
print("Gráfico guardado en Tema10/graficos/peticiones_series.png")
```

## Barras

Script `barras.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

incidencias = pd.read_csv("Tema10/datos/incidencias.csv")

# número de incidencias por servicio, de mayor a menor
por_servicio = incidencias["servicio"].value_counts()
print(por_servicio)

# barras verticales
figura, ejes = plt.subplots(figsize=(8, 4))
ejes.bar(por_servicio.index, por_servicio.values)
ejes.set_title("Incidencias por servicio")
ejes.set_xlabel("Servicio")
ejes.set_ylabel("Incidencias")
figura.savefig("Tema10/graficos/incidencias_servicio.png", dpi=150,
               bbox_inches="tight")

# minutos de interrupción por servicio, de menor a mayor
minutos = incidencias.groupby("servicio")["minutos_interrupcion"].sum()
minutos = minutos.sort_values()
print(minutos)

# barras horizontales: barh() deja los nombres legibles
figura, ejes = plt.subplots(figsize=(8, 4))
ejes.barh(minutos.index, minutos.values)
ejes.set_title("Minutos de interrupción por servicio")
ejes.set_xlabel("Minutos de interrupción")
figura.savefig("Tema10/graficos/minutos_servicio.png", dpi=150, bbox_inches="tight")
print("Gráficos guardados en Tema10/graficos")
```

## Dispersión

Script `dispersion.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

copias = pd.read_csv("Tema10/datos/copias.csv")

# se separan las copias por tipo con una máscara booleana
completas = copias[copias["tipo"] == "completa"]
incrementales = copias[copias["tipo"] == "incremental"]
print(len(completas), "completas y", len(incrementales), "incrementales")

figura, ejes = plt.subplots(figsize=(8, 4))
ejes.scatter(completas["tamano_gb"], completas["duracion_minutos"],
             color="#D1495B", marker="s", label="Completa")
ejes.scatter(incrementales["tamano_gb"], incrementales["duracion_minutos"],
             color="#2E86AB", marker="o", label="Incremental")
ejes.set_title("Tamaño y duración de las copias")
ejes.set_xlabel("Tamaño (GB)")
ejes.set_ylabel("Duración (minutos)")
ejes.grid(True)
ejes.legend()

figura.savefig("Tema10/graficos/copias_dispersion.png", dpi=150, bbox_inches="tight")
print("Gráfico guardado en Tema10/graficos/copias_dispersion.png")
```

## Fechas en el eje

Script `fechas.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

peticiones = pd.read_csv("Tema10/datos/peticiones.csv")
print(peticiones["fecha"].dtype)

# con fechas en texto, Matplotlib dibuja una etiqueta por cada valor
figura, ejes = plt.subplots(figsize=(8, 4))
ejes.plot(peticiones["fecha"], peticiones["peticiones_web"])
ejes.set_title("Fechas como texto")
figura.savefig("Tema10/graficos/fechas_texto.png", dpi=150, bbox_inches="tight")

# con fechas convertidas, elige sola las marcas del eje
peticiones["fecha"] = pd.to_datetime(peticiones["fecha"])
print(peticiones["fecha"].dtype)

figura, ejes = plt.subplots(figsize=(8, 4))
ejes.plot(peticiones["fecha"], peticiones["peticiones_web"])
ejes.set_title("Fechas convertidas")
ejes.tick_params(axis="x", rotation=45)  # etiquetas inclinadas
figura.savefig("Tema10/graficos/fechas_convertidas.png", dpi=150,
               bbox_inches="tight")
print("Gráficos guardados en Tema10/graficos")
```

## Panel con dos gráficos

Script `panel.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

peticiones = pd.read_csv("Tema10/datos/peticiones.csv")
peticiones["fecha"] = pd.to_datetime(peticiones["fecha"])
incidencias = pd.read_csv("Tema10/datos/incidencias.csv")
por_tipo = incidencias["tipo"].value_counts()

# subplots(1, 2) crea una fila con dos ejes; ejes es una lista de dos
figura, ejes = plt.subplots(1, 2, figsize=(10, 4))

ejes[0].plot(peticiones["fecha"], peticiones["peticiones_web"])
ejes[0].set_title("Peticiones web por día")
ejes[0].tick_params(axis="x", rotation=45)

ejes[1].bar(por_tipo.index, por_tipo.values)
ejes[1].set_title("Incidencias por tipo")

figura.tight_layout()  # evita que se solapen los textos
figura.savefig("Tema10/graficos/panel.png", dpi=150, bbox_inches="tight")
print("Panel guardado en Tema10/graficos/panel.png")
```

## Filtrar y dibujar

Script `graficos_graves.py`.

```python
import matplotlib.pyplot as plt
import pandas as pd

incidencias = pd.read_csv("Tema10/datos/incidencias.csv")

# cada condición va entre paréntesis para poder combinarlas con &
servicios = ["Correo", "Copias", "VPN", "Web"]
graves = incidencias[
    (incidencias["gravedad"] == "Alta")
    & (incidencias["servicio"].isin(servicios))
]
print(len(graves), "incidencias graves")

# se cuentan antes de dibujar
por_servicio = graves["servicio"].value_counts()
print(por_servicio)

figura, ejes = plt.subplots(figsize=(8, 4))
ejes.bar(por_servicio.index, por_servicio.values)
ejes.set_title("Incidencias de gravedad Alta por servicio")
ejes.set_xlabel("Servicio")
ejes.set_ylabel("Incidencias")

figura.savefig("Tema10/graficos/graves_servicio.png", dpi=150, bbox_inches="tight")
print("Gráfico guardado en Tema10/graficos/graves_servicio.png")
```
